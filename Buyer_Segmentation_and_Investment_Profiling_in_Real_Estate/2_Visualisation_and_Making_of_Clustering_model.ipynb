### Dropping Columns

```python
data = df.copy()
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
data.drop(columns=['client_id','price_segment', 'size_segment', 'age_segment', 'transaction_date','date_of_birth','transaction_quarters','transaction_month','transaction_day','tower_number','listing_status'], inplace=True)
```

Observation : <br>
Above we removed the unnecessary features like segments coz they were the features which were made from another feature. And we also have removed feature `listing_status` coz as we removed all the rows containing null values, thus the rows that contained information for units not being sold were also removed. And hence the feature `listing_status` now only contained the values of property being sold. Since there was only values being left in the entire column, so it was a wise idea to drop the entire column.

```python
data = data.dropna()
```

```python
data.isna().sum()
```

```python
data.columns
```

```python
with pd.option_context('display.max_columns', None):
    display(data.head())
```

---

### Apllying Scaling and Encoding
- Scaling : Here we choose Standard scaler because there is no min and max limit to the `floor_area_sqft` and `sale_price`
    - For features like `satisfaction_score` MinMax Scaler will be used
- Encoding : Here we are choosing One Hot Encoding for `unit_category` and `listing_status`

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer
```

Using Frequency Encode the high-cardinality columns

```python
# We map the country and region to how often they appear in the data
data['country_freq'] = data['country'].map(data['country'].value_counts())
data['region_freq'] = data['region'].map(data['region'].value_counts())
```

```python
# Notice that we added 'country_freq' and 'region_freq' here as numbers!
num_cols1 = ['sale_price', 'floor_area_sqft', 'age', 'country_freq', 'region_freq']
num_cols2 = ['satisfaction_score']

# Notice that we REMOVED the original 'country' and 'region' from here!
cat_cols = ['unit_category', 'client_type', 'gender', 
            'acquisition_purpose', 'loan_applied', 'referral_channel']
```

```python
transform = ColumnTransformer(transformers=[
    ('num1', StandardScaler(), num_cols1),
    ('num2', MinMaxScaler(), num_cols2),
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_cols)
])
```

```python
data_processed = transform.fit_transform(data)
```

```python
print(data_processed.shape)

data_processed
```

---

### Using KMeans for Clustering
Here we are directly using KMeans because we already know how many cluster (K) of buyer types we want. So instead of using Hierarchical Clustering we will proceed with KMeans Clustering.

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(
    n_clusters=4,
    init='k-means++',
    n_init=10,
    random_state=42
)
```

```python
prediction = model.fit_predict(data_processed)
```

```python
print(f"UNIQUE VALUES IN prediction : {np.unique(prediction)}")
```

Observation : <br>
Since we dont know which number belongs to which cluster type of buyers, so we need to put it with the dataset to recognise the different buyer type clusters. This process is called **`Cluster Profiling`**

---

### Putting The Predictions Back Into The Dataset

```python
sold_units = df['listing_status'] == 'Sold'

df.loc[sold_units, 'buyers_type'] = prediction
```

Replacing the `NaN` values in feature `buyers_type`

```python
df['buyers_type'] = df['buyers_type'].fillna('NO BUYERS')
```

---

```python
with pd.option_context('display.max_columns', None):
    display(df)
```

```python
df[['satisfaction_score', 'age']] = df[['satisfaction_score', 'age']].astype('Int64')
```

Observation :

**Cluster Profiling**

After running the K-Means algorithm, we analyzed the statistical averages of each group to assign them to our target Business Segments:

* **Cluster 3 = Luxury Investors (C4)**
  * This cluster has the highest average `sale_price` ($492,615) and the largest `floor_area_sqft` (1,612 sqft). They also have the highest satisfaction score (3.07). This perfectly matches the "High satisfaction, large investments" criteria!
  
* **Cluster 2 = First-Time Buyers (C2)**
  * This cluster has the absolute youngest average `age` (54.3 years old) across the entire dataset, and they have one of the highest `loan_applied` rates. This perfectly matches the "Younger, loan dependent" criteria!
  
* **Cluster 0 = Global Investors (C1)**
  * This cluster targets standard-sized apartments (avg 892 sqft) for solid rental yield investments.
  
* **Cluster 1 = Corporate Buyers (C3)**
  * This cluster targets compact units (avg 857 sqft), which are highly characteristic of companies buying smaller units for corporate housing or short-term employee stays.

### Renaming The Cluster Values

```python
val = {
    0 : 'Global Investors',
    1 : 'Corporate Buyers',
    2 : 'First Time Buyers',
    3 : 'Luxury Investors'
}

df['buyers_type'] = df['buyers_type'].map(val)
```
