#### FEATURE EXTRACTION -- Finding Age Of The Client From DOB
First we have to convert the dates into datetime format

```python
df.dtypes.reset_index()
```

Converting `transaction_date` and `date_of_birth` into datetime format

```python
df['transaction_date'] = df['transaction_date'].str.replace('-', '/')
df['date_of_birth'] = df['date_of_birth'].str.replace('-', '/')
```

```python
# The `errors='coerce'` argument turns invalid dates into NaT (Not a Time)
df['transaction_date'] = pd.to_datetime(df['transaction_date'], format='%m/%d/%Y', errors='coerce')
df['date_of_birth'] = pd.to_datetime(df['date_of_birth'], format='%m/%d/%Y', errors='coerce')
```

```python
df[['transaction_date', 'date_of_birth']].dtypes.reset_index()
```

```python
# Calculating Current Date
current_date = pd.to_datetime('today')

# Calculating Age
df['age'] = np.floor((current_date - df['date_of_birth']).dt.days / 365.25)
```

### Making Three Cluster Of Price (low, mid, high)

```python
x = df[['sale_price']]
```

```python
# Importing matplotlib and seaborn

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(20,4))
sns.scatterplot(x)
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
# Since We Are Already Using Only One Feature, Hence There Is No Need To Use Scaling On It

wcss = []

for i in range(1,11):
    km = KMeans(n_clusters=i, init='k-means++', random_state=42)
    km.fit_predict(x)
    wcss.append(km.inertia_)
```

```python
print(wcss)
```

```python
plt.plot(range(1,11), wcss)
plt.title('Elobow Method Price Segmentation')
plt.xlabel('Clusters')
plt.ylabel('WCSS')
plt.show()
```

Observation : <br>
Since the optimal WCSS is at 2 and 3, we should go with having 3 clusters of price (Budget, Mid-market, Premium).

```python
km = KMeans(n_clusters=3, init='k-means++', random_state=42)
price_segment = km.fit_predict(x)
```

```python
df['price_segment'] = pd.DataFrame(price_segment)

print(df['price_segment'].unique())
```

```python
print(f"LABEL 0 :   MIN - {df[df['price_segment'] == 0]['sale_price'].min()},  MAX - {df[df['price_segment'] == 0]['sale_price'].max()}")
print(f"LABEL 1 :   MIN - {df[df['price_segment'] == 1]['sale_price'].min()},    MAX - {df[df['price_segment'] == 1]['sale_price'].max()}")
print(f"LABEL 2 :   MIN - {df[df['price_segment'] == 2]['sale_price'].min()},  MAX - {df[df['price_segment'] == 2]['sale_price'].max()}")
```

Now we will be assigning names to the labels : <br> 
- 1 - Budget
- 2 - Mid-market
- 0 - Premium

```python
segments = {
    1 : 'Budget',
    2 : 'Mid-market',
    0 : 'Premium'
}

df['price_segment'] = df['price_segment'].map(segments)
```

### Adding Three More Features From `transaction_date`
Making two more features such as `transaction_month`, `transaction_day`, and `transaction_quarters`

```python
month = df['transaction_date'].dt.month_name()
day_name = df['transaction_date'].dt.day_name()
quarters = df['transaction_date'].dt.quarter
```
