---

<h1 align='center'>Feature Labeling</h1>

`Education`

```python
edu = {
    1:"Below College", 
    2:"College", 
    3:"Bachelor", 
    4:"Master", 
    5:"Doctor"
}

df['Education Labels'] = df['Education'].map(edu)
```

`Environment Satisfaction`

```python
env_satisfaction = {
    1 : "Low", 
    2 : "Medium",
    3 : "High",
    4 : "Very High"
}

df['Environment Satisfaction Labels'] = df['Environment Satisfaction'].map(env_satisfaction)
```

`Job Involvement`

```python
job_involv = {
    1: "Minimally Involved",
    2: "Moderately Involved",
    3: "Highly Involved",
    4: "Fully Dedicated"
}

df['Job Involvement Labels'] = df['Job Involvement'].map(job_involv)
```

`Job Level`

```python
job_lvl = {
    1: 'Associate / Junior',
    2: 'Mid-Level',
    3: 'Senior',
    4: 'Staff / Principal',
    5: 'Director / VP'
}

df['Job Level Label'] = df['Job Level'].map(job_lvl)
```

`Relationship Satisfaction`

```python
relation_satisfaction = {
    1: 'Very Dissatisfied',
    2: 'Dissatisfied',
    3: 'Satisfied',
    4: 'Very Satisfied'
}

df['Relationship Satisfaction Labels'] = df['Relationship Satisfaction'].map(relation_satisfaction)
```

`Total Working Years`

```python
num_companies = df['Num Companies Worked'].replace(0, 0.5)
df['Avg Tenure Per Company'] = (df['Total Working Years'] / num_companies).round(1)
```

```python
bins = [-1, 2.0, 5.0, float('inf')]
labels = [
    'Low Stability (< 2 yrs)',
    'Moderate Stability (2-5 yrs)',
    'High Stability (5+ yrs)'
]

df['Candidate Stability'] = pd.cut(df['Avg Tenure Per Company'],bins=bins,labels=labels,ordered=True)
```

`Training Times Last Year`

```python
tenure = df['Years At Company'].replace(0, 0.5)

df['Training Intensity'] = (df['Training Times Last Year'] / tenure).round(2)
```

`Work Life Balance`

```python
work_life = {
    1 : 'Poor',
    2 : 'Fair',
    3 : 'Good',
    4 : 'Excellent'
}

df['Work Life Balance Labels'] = df['Work Life Balance'].map(work_life)
```

`Years In Current Company`

```python
df['Role Stagnation Index'] = (df['Years In Current Role'] / tenure).round(2)
```

`Years Since Last Promotion`

```python
df['Promotion Gap Ratio'] = (df['Years Since Last Promotion'] / tenure).round(2)
```

`Years With Curr Manager`

```python
bins = [-1,1,4,9, df['Years With Curr Manager'].max()]

labels1 = [1,2,3,4]

labels2 = [
    'New Relationship',
    'Developing',
    'Stable',
    'Highly Stable'
]

df['Manager Stability Indicator'] = pd.cut(df['Years With Curr Manager'], bins=bins, labels=labels1)
df['Manager Stability Indicator Labels'] = pd.cut(df['Years With Curr Manager'], bins=bins, labels=labels2)
```
