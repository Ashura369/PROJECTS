<h2 align='center'>Applying K-Means</h2>

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
clusters = kmeans.fit_predict(df)
```

---

<h2 align='center'>Plotting the Clusters</h2>

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
clusters_pca = pca.fit_transform(df)
```

```python
df_plot = pd.DataFrame(clusters_pca, columns=['1','2'])
df_plot['Clusters'] = clusters

plt.figure(figsize=(10,6))
sns.scatterplot(x='1',y='2',hue='Clusters', palette='viridis', data=df_plot)
plt.suptitle('PLOTTING THE FOUR DIFFERENT CLUSTERS')
plt.show()
```

<h2 align='center'>Renaming the Clusters</h2>


After analyzing the dataset it is observed that, the clusters should be renamed as :
- 0: 'Tenured & Stalled Seniors',
- 1: 'Fast-Track High Performers',
- 2: 'Early-Career Explorers',
- 3: 'Role-Stagnant Mid-Level'

```python
rename = {
    0: 'Tenured & Stalled Seniors',
    1: 'Fast-Track High Performers',
    2: 'Early-Career Explorers',
    3: 'Role-Stagnant Mid-Level'
}

data2['Promotion Risk Cluster'] = data2['Promotion Risk Cluster'].map(rename)
```
