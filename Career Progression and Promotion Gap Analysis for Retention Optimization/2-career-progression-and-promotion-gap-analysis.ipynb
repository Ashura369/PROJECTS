<h2 align='center'>Applying Robust Scaling on Outliers</h2>

```python
from sklearn.preprocessing import RobustScaler
```

```python
features = ['Monthly Income','Years Since Last Promotion','Years At Company']
features_scaled = [f"{i}_Scaled" for i in features]

robust_scaler = RobustScaler() 
df[features_scaled] = robust_scaler.fit_transform(df[features])
```

```python
df.drop(columns=features, inplace=True)
```

<h2 align='center'>Applying Standard Scaler</h2>

```python
df.drop(columns=features2, inplace=True)
```
